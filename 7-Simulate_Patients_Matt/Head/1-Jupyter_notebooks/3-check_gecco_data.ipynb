```python

import os
import numpy as np

def load_npy_files_to_dict(directory):
    data_dict = {}

    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith('.npy'):
                array = np.load(os.path.join(root, file))
                data_dict[os.path.splitext(file)[0]] = array  # Use file name as key

    return data_dict

data_dict = load_npy_files_to_dict('./gecco_data')
```

```python
import matplotlib.pyplot as plt
import numpy as np

def process_dict(data_dict):
    for key, array in data_dict.items():
        print(f"Key: {key}")
        print(f"Mean: {np.mean(array)}")
        print(f"Max: {np.max(array)}")
        print(f"Min: {np.min(array)}")

        plt.figure()
        if array.ndim == 1:
            plt.plot(array)
        elif array.ndim == 2:
            plt.imshow(array)
        elif array.ndim == 3:
            # For 3D arrays, plot a slice in the middle
            plt.imshow(array[array.shape[0] // 2],cmap='turbo',vmin=0,vmax=np.percentile(array[0],50))
        plt.title(key)
        plt.show()

process_dict(data_dict)
```
