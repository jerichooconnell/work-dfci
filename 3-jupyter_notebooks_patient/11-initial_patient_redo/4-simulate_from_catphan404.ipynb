```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.subplot(131)

from scipy.ndimage import gaussian_filter

filt_sigma = 1

int1 = gaussian_filter(phantom.ggems_primary_projections[0],filt_sigma)
int2 = gaussian_filter(phantom.primary_projections[0],filt_sigma)

proj1 = -np.log((np.abs(int1+0.001))/np.rot90(phantom.ggems_flood + 0.001))
proj2 = -1.004*np.log((int2)/phantom.flood_field) # 1.12*
# gaussian blur the projections using scipy ndimage 

crop = 5

plt.imshow(np.flipud(proj1[crop:-crop,crop:-crop]),cmap='turbo',vmin=0,vmax=4.5)
plt.axis('off')
plt.title(f'a) $\mu$ GGEMS [cm$^{-1}$]')
plt.colorbar(orientation='horizontal')
plt.subplot(132)
plt.imshow(np.flipud(proj2[crop:-crop,crop:-crop]),cmap='turbo',vmin=0,vmax=4.5)
plt.axis('off')
plt.title(f'b) $\mu$ GECCO [cm$^{-1}$]')
plt.colorbar(orientation='horizontal')

# Plot the difference 
plt.subplot(133)
diff_1 = 100*((proj1[crop:-crop,crop:-crop]-proj2[crop:-crop,crop:-crop])/np.max(proj2[crop:-crop,crop:-crop]))

diff_filt = gaussian_filter(diff_1,1)
plt.axis('off')
plt.title(f'c) Difference [%]')
plt.imshow(np.flipud(diff_filt),cmap='bwr',vmin=-1,vmax=1)
plt.colorbar(orientation='horizontal')

plt.tight_layout()
plt.savefig('Catphan404_comparison.png',dpi=300)
# plt.savefig('XCAT_comparison.png',dpi=300)
# plot the mean of the difference image along both axes
plt.figure()
plt.subplot(121)
plt.plot(np.mean(phantom.flood_field,axis=0))
plt.plot(np.mean(phantom.ggems_flood,axis=1))

plt.subplot(122)
plt.plot(np.mean(proj1,axis=0))
plt.plot(np.mean(proj2,axis=0))

np.mean(np.abs(diff_filt))


```
